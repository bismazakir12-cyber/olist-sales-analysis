## Findings
- The orders table has 99,441 rows and covers 2016-09-04 to 2018-10-17.
- Most orders are delivered (about 97.0%). 625 orders were canceled, and a small number are shipped, unavailable, invoiced or still processing.
- The customer delivery date is missing for 2,965 orders. Almost all of these were not delivered, which is expected.
- Review scores are mostly 5 stars (57,328 of 99,224 reviews, about 57.8%). 1 star is the second most common score (11,424).
- There are 96,096 unique customers but 99,441 customer IDs, because customer_id changes for every order.
